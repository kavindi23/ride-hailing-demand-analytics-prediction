```python
import pandas as pd

columns_to_check = [
    "request_datetime",
    "pickup_datetime",
    "dropoff_datetime",
    "PULocationID",
    "DOLocationID",
    "trip_miles",
    "trip_time",
    "base_passenger_fare",
    "tips",
    "driver_pay"
]

total_rows = 0
missing_counts = {col: 0 for col in columns_to_check}

zero_distance = 0
negative_distance = 0

zero_trip_time = 0
negative_trip_time = 0

negative_fare = 0

for batch in parquet_file.iter_batches(
    batch_size=500_000,
    columns=columns_to_check
):
    df_batch = batch.to_pandas()

    total_rows += len(df_batch)

    # Count missing values
    for col in columns_to_check:
        missing_counts[col] += df_batch[col].isna().sum()

    # Check distance
    zero_distance += (df_batch["trip_miles"] == 0).sum()
    negative_distance += (df_batch["trip_miles"] < 0).sum()

    # Check trip duration
    zero_trip_time += (df_batch["trip_time"] == 0).sum()
    negative_trip_time += (df_batch["trip_time"] < 0).sum()

    # Check fare
    negative_fare += (df_batch["base_passenger_fare"] < 0).sum()

print("Total rows:", total_rows)

print("\nMissing values:")
for col, count in missing_counts.items():
    percentage = (count / total_rows) * 100
    print(f"{col}: {count:,} ({percentage:.2f}%)")

print("\nPotential data-quality issues:")
print(f"Zero trip distance: {zero_distance:,}")
print(f"Negative trip distance: {negative_distance:,}")
print(f"Zero trip time: {zero_trip_time:,}")
print(f"Negative trip time: {negative_trip_time:,}")
print(f"Negative base fare: {negative_fare:,}")
```

```python
columns_to_check = [
    "request_datetime",
    "pickup_datetime",
    "dropoff_datetime",
    "PULocationID",
    "DOLocationID",
    "trip_miles",
    "trip_time",
    "base_passenger_fare",
    "tips",
    "driver_pay"
]

stats = {
    "min_request": None,
    "max_request": None,
    "min_distance": float("inf"),
    "max_distance": float("-inf"),
    "min_trip_time": float("inf"),
    "max_trip_time": float("-inf"),
    "min_fare": float("inf"),
    "max_fare": float("-inf"),
    "min_tip": float("inf"),
    "max_tip": float("-inf"),
    "min_driver_pay": float("inf"),
    "max_driver_pay": float("-inf")
}

pickup_before_request = 0
dropoff_before_pickup = 0

pickup_locations = set()
dropoff_locations = set()

for batch in parquet_file.iter_batches(
    batch_size=500_000,
    columns=columns_to_check
):
    df = batch.to_pandas()

    # Date range
    batch_min_request = df["request_datetime"].min()
    batch_max_request = df["request_datetime"].max()

    if stats["min_request"] is None or batch_min_request < stats["min_request"]:
        stats["min_request"] = batch_min_request

    if stats["max_request"] is None or batch_max_request > stats["max_request"]:
        stats["max_request"] = batch_max_request

    # Numerical ranges
    stats["min_distance"] = min(stats["min_distance"], df["trip_miles"].min())
    stats["max_distance"] = max(stats["max_distance"], df["trip_miles"].max())

    stats["min_trip_time"] = min(stats["min_trip_time"], df["trip_time"].min())
    stats["max_trip_time"] = max(stats["max_trip_time"], df["trip_time"].max())

    stats["min_fare"] = min(stats["min_fare"], df["base_passenger_fare"].min())
    stats["max_fare"] = max(stats["max_fare"], df["base_passenger_fare"].max())

    stats["min_tip"] = min(stats["min_tip"], df["tips"].min())
    stats["max_tip"] = max(stats["max_tip"], df["tips"].max())

    stats["min_driver_pay"] = min(stats["min_driver_pay"], df["driver_pay"].min())
    stats["max_driver_pay"] = max(stats["max_driver_pay"], df["driver_pay"].max())

    # Logical timestamp checks
    pickup_before_request += (
        df["pickup_datetime"] < df["request_datetime"]
    ).sum()

    dropoff_before_pickup += (
        df["dropoff_datetime"] < df["pickup_datetime"]
    ).sum()

    # Location IDs
    pickup_locations.update(df["PULocationID"].unique())
    dropoff_locations.update(df["DOLocationID"].unique())

print("REQUEST DATE RANGE")
print("Minimum:", stats["min_request"])
print("Maximum:", stats["max_request"])

print("\nNUMERICAL RANGES")
print("Trip miles:", stats["min_distance"], "to", stats["max_distance"])
print("Trip time:", stats["min_trip_time"], "to", stats["max_trip_time"])
print("Base fare:", stats["min_fare"], "to", stats["max_fare"])
print("Tips:", stats["min_tip"], "to", stats["max_tip"])
print("Driver pay:", stats["min_driver_pay"], "to", stats["max_driver_pay"])

print("\nTIMESTAMP ISSUES")
print("Pickup before request:", pickup_before_request)
print("Dropoff before pickup:", dropoff_before_pickup)

print("\nLOCATION IDs")
print("Pickup ID range:", min(pickup_locations), "to", max(pickup_locations))
print("Dropoff ID range:", min(dropoff_locations), "to", max(dropoff_locations))
print("Unique pickup locations:", len(pickup_locations))
print("Unique dropoff locations:", len(dropoff_locations))
```

```python
import numpy as np

sample_parts = []

for batch in parquet_file.iter_batches(
    batch_size=500_000,
    columns=[
        "trip_miles",
        "trip_time",
        "base_passenger_fare",
        "tips",
        "driver_pay"
    ]
):
    df = batch.to_pandas()

    # Take a reproducible sample from every batch
    sample_parts.append(
        df.sample(
            n=min(10_000, len(df)),
            random_state=42
        )
    )

distribution_sample = pd.concat(
    sample_parts,
    ignore_index=True
)

print("Sample size:", len(distribution_sample))

distribution_sample.describe(
    percentiles=[0.5, 0.90, 0.95, 0.99, 0.999]
).T
```
