SQL Insight: Ride demand varies substantially by date and hour. January 2 recorded the lowest daily demand, while 6 PM was the highest-demand hour overall. Although weekdays generated more total rides because there were more weekday observations, average zone-hour demand was higher on weekends (109.16) than on weekdays (102.58).

SQL Insight: Among zones averaging at least 100 pickups per hour, Lower East Side showed the highest relative demand variability, with a coefficient of variation of 0.830. Several Manhattan zones also exhibited substantial variability, while LaGuardia Airport combined very high average hourly demand with considerable variation. These results suggest that demand-prediction difficulty may differ across high-volume locations.

SQL Insight: The highest-demand pickup zone differed across boroughs. JFK Airport recorded the highest demand in Queens, Crown Heights North led Brooklyn, East Village led Manhattan, Mott Haven/Port Morris led the Bronx, and Saint George/New Brighton recorded the highest demand in Staten Island. This demonstrates that ride-demand hotspots vary geographically and supports the use of location-level features in demand prediction.

SQL Insight: Peak ride-demand hours vary geographically. Manhattan and Brooklyn recorded their highest demand at 6 PM, while Queens, the Bronx, and Staten Island peaked at 8 AM. This indicates that temporal demand patterns differ across locations, suggesting that both pickup location and time-of-day—and their interaction—are important features for demand prediction.

SQL Insight: Peak ride-demand hours vary across the week. Monday, Tuesday, and Thursday reached their highest demand at 8 AM, while Wednesday and Friday peaked at 6 PM. Saturday peaked later at 7 PM, and Sunday recorded its highest demand at midnight. This demonstrates that hourly demand patterns depend on the day of the week, making both time-of-day and day-of-week important features for demand prediction.
